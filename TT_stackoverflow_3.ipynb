```python
# Remote, Hybrid (some in person, flexibility), In-person, Hybrid (some remote, heavy in person), Your Choice
```

Dataframe con los encuestados que afirmaron trabajar en remoto o significativamente mas en remoto. Se excluyen las filas donde la columna convertedcompyearly almacena un valor nulo.

```python
mask_remote = ((df['remotework'] == 'Remote') | (df['remotework'] == 'Hybrid (some in-person, leans heavy to flexibility)') | (df['remotework'] == 'Your choice (very flexible, you can come in when you want or just as needed)')) & (df['convertedcompyearly'].notna())
df_remote = df.loc[mask_remote][['remotework', 'convertedcompyearly']]
```

```python
df_remote.shape[0]
```

```python
remote_total = df_remote.groupby('remotework')['convertedcompyearly'].agg(['median', 'count']).sort_values(by='median', ascending=False)
remote_total
```

```python
print(f'Encuestados que afirmaron trabajar en remoto: {remote_total.iloc[0]['count'].sum()}.')
print(f'Salario medio anual de los encuestados que trabajan remoto: {remote_total.iloc[0]['median']} dolares.')
print(f'Encuestados que afirmaron trabajar en remoto o significativamente mas en remoto: {remote_total['count'].sum()}.')
```

Dataframe con los datos de los encuestados que afirmaron trabajar presencial o significativamente mas en presencial. Se excluyen las filas donde la columna convertedcompyearly almacena un valor nulo.

```python
mask_in_person = ((df['remotework'] == 'In-person') | (df['remotework'] == 'Hybrid (some remote, leans heavy to in-person)')) & (df['convertedcompyearly'].notna())
df_in_person = df.loc[mask_in_person][['remotework', 'convertedcompyearly']]
```

```python
df_in_person.shape[0]
```

```python
in_person_total = df_in_person.groupby('remotework')['convertedcompyearly'].agg(['median', 'count']).sort_values(by='median', ascending=False)
in_person_total
```

```python
print(f'Encuestados que afirmaron trabajar presencial: {in_person_total.iloc[1]['count'].sum()}')
print(f'Salario medio anual de los encuestados que trabajan presencial: {in_person_total.iloc[1]['median']} dolares.')
print(f'Encuestados que afirmaron trabajar presencial o significativamente mas en regimen presencial: {in_person_total['count'].sum()}.')
```

Se obtiene el salario medio para ambos grupos de encuestados, quienes afirmaron trabajar mas en remoto y quienes afirmaron trabajar mas en presencial.

```python
remote_salary = df_remote['convertedcompyearly'].median()
remote_salary
```

```python
in_person_salary = df_in_person['convertedcompyearly'].median()
in_person_salary
```

```python
print(f'Un desarrollador que trabaja remoto o significativamente mas en remoto gana {remote_salary} dolares anuales.')
print(f'Un desarrollador que trabaja presencial o significativamente mas presencial gana {in_person_salary} dolares anuales.')
print(f'La diferencia es de {remote_salary - in_person_salary} dolares anuales en favor de los desarrolladores que trabajan significativamente mas en remoto.')
```
