```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/encuestas_tinder.csv')
pd.set_option('display.max_columns', None)
```

```python
df.shape
```

```python
df.info(memory_usage='deep')
```

```python
#for c in df.select_dtypes('object'):
#  df[c] = df[c].astype('object')
```

```python
for i in df.select_dtypes('integer'):
  df[i] = pd.to_numeric(df[i], downcast='integer')
```

```python
for f in df.select_dtypes('float'):
  df[f] = pd.to_numeric(df[f], downcast='float')
```

```python
df.info(memory_usage='deep')
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(' ', '_')
)
df.columns
```

```python
nulls_count = df.isnull().sum()
nulls_count[nulls_count > 0].sort_values(ascending=False)
```

```python
nulls_percentage = df.isnull().mean()*100
nulls_percentage[nulls_percentage > 0].sort_values(ascending=False)
```

```python
duplicates_sum = df.duplicated().sum()
duplicates_sum
```

```python
duplicates_percentage = (df.duplicated().sum()/df.shape[0]) * 100
duplicates_percentage
```

```python
show_duplicates = df[df.duplicated(keep=False)].sort_values(by=list(df.columns)).head()
show_duplicates
```

**¿Que sexo consigue mas citas?**

```python
df.columns
```

```python
df.head()
```

```python
df['gender'].unique()
```

```python
df['answer'].value_counts()
```

```python
mask = df['answer'] == 'Yes'
tinder_date = df.loc[mask]
tinder_date.shape
```

```python
tinder_date.head()
```

```python
tinder_date['gender'].value_counts()
```

```python
tinder_date['gender'].value_counts(normalize=True)
```

```python
date_by_sex = pd.DataFrame(tinder_date['gender'].value_counts()).reset_index()
date_by_sex.rename(columns={'gender': 'sex', 'count': 'dates'}, inplace=True)
date_by_sex['%'] = round((date_by_sex['dates'] / date_by_sex['dates'].sum()) * 100, 2)
```

```python
date_by_sex
```

**¿Que sexo consigue mas parejas?**

```python
mask = (df['answer'] == 'Yes') & (df['it_became_a_relationship'] == 'Yes')
relationship = df.loc[mask]
relationship.shape
```

```python
relationship.head()
```

```python
relationship['gender'].value_counts(normalize=True)
```

```python
relationship_by_sex = pd.DataFrame(relationship['gender'].value_counts()).reset_index()
relationship_by_sex.rename(columns={'gender': 'sex', 'count': 'total_relationships'}, inplace=True)
relationship_by_sex['%'] = round((relationship_by_sex['total_relationships'] / relationship_by_sex['total_relationships'].sum()) * 100, 2)
relationship_by_sex
```
