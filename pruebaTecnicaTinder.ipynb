**¿Que Sexo consigue mas parejas en la plataforma Tinder?**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/encuestas_tinder.csv')
df.head()
```

Se filtran las filas donde el valor para la columna 'Answer' es Yes y donde el valor de la columna 'It became a Relationship' es Yes, es decir, solo los casos de exito.

```python
mask = (df['Answer'] == 'Yes') & (df['It became a Relationship'] == 'Yes')
df_sucess = df[mask]
df_sucess.shape[0]
```

```python
df_sucess.head()
```

Cantidad de exitos por Sexo.

```python
df_sucess['Gender'].value_counts()
```

Se crea un nuevo DataFrame con los datos obtenidos en el paso anterior.

```python
sucess_rate_per_sex = pd.DataFrame(df_sucess['Gender'].value_counts())
sucess_rate_per_sex.head()
```

Se crea una nueva columna para el DataFrame creado que incluya el porcentaje de exito.

```python
sucess_rate_per_sex['%'] = round(sucess_rate_per_sex['count'] / sucess_rate_per_sex['count'].sum() * 100, 2)
sucess_rate_per_sex.head()
```

**¿Cual es el porcentaje de personas que consiguieron pareja usando Tinder?**

```python
df.columns
```

Filtrar los datos de las personas que usan la plataforma (columna Answer) y obtener el porcentaje de las personas que consiguieron pareja (columna 'It became a Relationship')

```python
usuarios_tinder = df.groupby('Answer')['It became a Relationship'].value_counts(normalize=True)
usuarios_tinder
```

Cantidad de encuestados que utilizan Tinder

```python
tinder_users = df[df['Answer'] == 'Yes']
tinder_users.shape[0]
```

```python
tinder_users['It became a Relationship'].value_counts()
```

```python
tinder_sucess_rate = pd.DataFrame(tinder_users['It became a Relationship'].value_counts())
tinder_sucess_rate['%'] = round(tinder_sucess_rate['count'] / tinder_sucess_rate['count'].sum() * 100, 2)
tinder_sucess_rate
```

**Tomando en cuenta la totalidad de los encuestados (quienes usan y quienes no la plataforma), obtener el porcentaje de los mismos que consiguieron pareja usando Tinder**

```python
mask = (df['Answer'] == 'Yes') & (df['It became a Relationship'] == 'Yes')
sucess = df[mask]
sucess.shape[0]
```

```python
sucess_percentage = round((sucess.shape[0]) / (df.shape[0]) * 100, 2)
f'El porcentaje de encuestados que consiguieron pareja usando Tinder es: {sucess_percentage}%'
```
