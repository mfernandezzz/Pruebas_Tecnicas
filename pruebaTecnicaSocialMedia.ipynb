**¿Con que edad se consumen mas las redes sociales?**

Obtener el promedio de edad y de cantidad de horas de uso de las redes sociales. Tomar como referencia los valores de la columna Daily_Screen_Time_Hours y la columna edad.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/social_media_mental_health.csv')
df.head()
```

```python
df.shape[0]
```

```python
df['User_ID'].duplicated().value_counts()
```

Se agrupa por edad y se calcula el tiempo de consumo diario

```python
result = df.groupby('Age')['Daily_Screen_Time_Hours'].mean()
```

```python
result.idxmax()
```

```python
round(result.max(), 1)
```

```python
f'La edad en la que se consumen mas redes sociales es {result.idxmax()} años y el promedio de horas diario es de {round(result.max(), 1)} horas'
```
