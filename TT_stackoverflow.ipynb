Otra solucion

```python
langs = ['Python', 'SQL', 'JavaScript', 'Rust', 'TypeScript']
```

```python
df = df[(df['convertedcompyearly'].notna())]
df.shape
```

```python
salarios = {}
for lang in langs:
  mask = df['languagehaveworkedwith'].str.contains(lang, na=False)

  salarios[lang] = (df[mask]['convertedcompyearly'].median())

  result = pd.Series(salarios)
  result = result.sort_values(ascending=False)

print('Salario medio por lenguage: ')
print(result.round(0))
```
