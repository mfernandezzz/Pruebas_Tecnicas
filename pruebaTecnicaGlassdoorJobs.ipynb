**¿Cual es el salario promedio anual de un Data Engineer?**

```python
data_engineer = df.loc[(df['Job Title'] == 'Data Engineer') & (df['Salary Estimate'] != '-1')]
data_engineer.shape
```

```python
data_engineer['Salary Estimate'].unique()
```

```python
data_engineer['Salary Estimate'] = (
    data_engineer['Salary Estimate'].astype(str)
    .str.strip()
    .str.replace('(Glassdoor est.)', '')
    .str.replace('Employer Provided Salary:', '')
    .str.replace('Employer est.', '')
    .str.replace('K', '000')
    .str.replace('$', '')
    .str.replace('()', '')
    .str.replace(' ', '')
    )

data_engineer['Salary Estimate'].unique()
```

```python
salaries = data_engineer['Salary Estimate'].str.split('-')
salaries.head()
```

```python
def min_max_values(salaries):
  min_salary, max_salary = [], []
  for s in salaries:
    min_salary.append(s[0])
    max_salary.append(s[1])
  return {
      'min_salaries': min_salary,
      'max_salaries': max_salary
  }

# min_max_values(salaries)

range_salaries = pd.DataFrame(min_max_values(salaries))
range_salaries.head()
```

```python
range_salaries.dtypes
```

```python
range_salaries['min_salaries'] = range_salaries['min_salaries'].astype(int)
range_salaries['max_salaries'] = range_salaries['max_salaries'].astype(int)
range_salaries.dtypes
```

```python
range_salaries['avg'] = ((range_salaries['min_salaries']) + (range_salaries['max_salaries'])) / 2
range_salaries.head()
```

```python
data_engineer_avg_salary = sum(range_salaries['avg']) / (range_salaries.shape[0])
f'El salario promedio anual de un Data Engineer es de {round(data_engineer_avg_salary, 2)} dolares.'
```

```python
# not accurate
avg_min_sal = range_salaries['min_salaries'].sum() / range_salaries.shape[0]
avg_max_sal = range_salaries['max_salaries'].sum() / range_salaries.shape[0]
xtwo = ((avg_min_sal) + (avg_max_sal) / 2)
xtwo
```
