**¿Cuanto gana en promedio un Data Enginner?**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/glassdoor_jobs_clean.csv')
df.head()
```

```python
df.info()
```

```python
#cantidad de registros correspondientes a cada profesion
df['Job Title'].value_counts()
```

```python
#se filtran las filas con los datos correspondientes a la profesion Data Enginner
#se filtran los resultados donde el valor para la columna Salary Estimate sea -1
mask = (df['Job Title'] == 'Data Engineer') & (df['Salary Estimate'] != '-1')
df_data_engineer = df[mask]
df_data_engineer.head(10)
```

```python
#se crea una Serie solo con los datos de la columna Salary Estimate
salaries = pd.Series(df_data_engineer['Salary Estimate'])
salaries
```

Los datos almcenados en la columna 'Salary Estimate' son strings, los cuales representan un rango salarial. Por consiguiente, se debe modificar dichos strings para que el dato almacenado en la Serie sea lo mas parecido a un rango numerico.

```python
salaries = salaries.str.replace('(Glassdoor est.)', '')
salaries = salaries.str.replace('K', '')
salaries = salaries.str.replace('$', '')
salaries = salaries.str.replace('Employer Provided Salary:', '')
salaries = salaries.str.replace('-', ',')
salaries = salaries.str.replace('(Employer est.)', '')
salaries = salaries.str.replace(' ', '')
salaries
```

```python
#se convierte la serie a una lista que almacena los salarios como string
salaries = salaries.to_list()
salaries
```

Se itera en la lista de strings para separarlos y transformarlos a valor numerico. El resultado es una lista que almacena listas con dos elementos numericos, donde el elemento en la posicion 0 de cada lista es el valor minimo y el elemento en la posicion 1 es el valor maximo.

```python
def get_num(salaries):
  #se dividen los strings y se los almacena en una lista
  salaries = [s.split(',') for s in salaries]
  #se itera en la lista para transformar los strings a valor numerico
  salaries = [[int(s[0]), int(s[1])] for s in salaries]
  return salaries

salaries = get_num(salaries)
salaries
```

Funcion que obtiene el valor promedio de los valores minimos

```python
def avg_min(salaries):
  min_salaries = [s[0] for s in salaries]
  return (sum(min_salaries) / len(min_salaries))

round(avg_min(salaries), 1)
```

Funcion que obtiene el valor promedio de los valores maximos

```python
def avg_max(salaries):
  max_salaries = [s[1] for s in salaries]
  return (sum(max_salaries) / len(max_salaries))

round(avg_max(salaries), 1)
```

Valor del salario promedio en base al promedio minimo y el promedio maximo

```python
avg_salary = (avg_min(salaries) + avg_max(salaries)) / 2
round(avg_salary, 1)
```
