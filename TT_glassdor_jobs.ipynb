**¿Cual es el salario promedio anual de un Data Engineer?**

```python
df.head()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df['job_title'].value_counts().head()
```

```python
mask = (df['job_title'] == 'Data Engineer') & (df['salary_estimate'] != '-1')
df_data_engineer = df.loc[mask]
df_data_engineer.shape
```

```python
df_data_engineer['salary_estimate'].unique()
```

```python
df_data_engineer['salary_estimate'] = (df_data_engineer['salary_estimate']
                                       .astype(str)
                                       .str.strip()
                                       .str.replace('(Glassdoor est.)', '')
                                       .str.replace('(Employer est.)', '')
                                       .str.replace('Employer Provided Salary:', '')
                                       .str.replace('$', '')
                                       .str.replace('K', '000')
                                       .str.replace(' ', ''))
df_data_engineer['salary_estimate'].unique()
```

```python
data_engineer_salaries = df_data_engineer['salary_estimate'].str.split('-')
data_engineer_salaries.head()
```

```python
def get_min_max_values(salaries):
  min_values = []
  max_values = []
  for sal in salaries:
    min_values.append(sal[0])
    max_values.append(sal[1])
  return {
      'min_values': min_values,
      'max_values': max_values
  }

data_engineer_salary_ranges = pd.DataFrame(get_min_max_values(data_engineer_salaries))
```

```python
data_engineer_salary_ranges.head()
```

```python
data_engineer_salary_ranges.dtypes
```

```python
data_engineer_salary_ranges['min_values'] = data_engineer_salary_ranges['min_values'].astype(int)
data_engineer_salary_ranges['max_values'] = data_engineer_salary_ranges['max_values'].astype(int)
```

```python
data_engineer_salary_ranges.dtypes
```

```python
data_engineer_salary_ranges['avg'] = ((data_engineer_salary_ranges['min_values']) + (data_engineer_salary_ranges['max_values'])) / 2
data_engineer_salary_ranges.head()
```

```python
data_engineer_anual_salary = data_engineer_salary_ranges['avg'].sum() / data_engineer_salary_ranges.shape[0]
```

```python
f'El salario promedio anual de un Data Engineer es de {round(data_engineer_anual_salary, 2)} dolares.'
```
